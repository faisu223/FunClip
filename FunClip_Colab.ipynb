# FunClip: One-Click Colab Notebook

This notebook will set up and run the FunClip application with all the latest features for you.

**Instructions:**
Just run the single code cell below. It will:
1.  Clone the correct version of the FunClip repository.
2.  Install all dependencies.
3.  Launch the Gradio application.

This process will take a few minutes. At the end, a public Gradio link will appear. Click it to use FunClip.

```python
# Step 1: Clone the repository with the new features
# We remove the old directory just in case you are re-running this cell
!rm -rf FunClip
!git clone --branch feature/add-gemini-srt-colab https://github.com/alibaba-damo-academy/FunClip.git
%cd FunClip

# Step 2: Install dependencies
!pip install -r requirements.txt
!apt-get -y update && apt-get -y install --fix-missing ffmpeg imagemagick
!sed -i 's/none/read,write/g' /etc/ImageMagick-6/policy.xml
!wget https://isv-data.oss-cn-hangzhou.aliyuncs.com/ics/MaaS/ClipVideo/STHeitiMedium.ttc -O font/STHeitiMedium.ttc
print("✅ Dependencies installed.")

# Step 3: Launch the application
print("🚀 Launching FunClip... please wait for the public URL to appear.")
!python funclip/launch.py --share
```
