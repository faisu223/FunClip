```python
# Step 1: Install dependencies (with numpy fix)
!pip install numpy>=2.0.0 --quiet
!pip install librosa soundfile scikit-learn>=1.3.2 funasr>=1.1.2 moviepy==1.0.3 gradio modelscope torch>=1.13 torchaudio openai g4f dashscope curl_cffi google-generativeai --quiet
!apt-get -y update && apt-get -y install --fix-missing ffmpeg imagemagick --quiet
!sed -i 's/none/read,write/g' /etc/ImageMagick-6/policy.xml
print("✅ Dependencies installed.")

# Step 2: Create and run a bootstrap script to write all application files
import os

file_contents = {}
# --- All file contents are embedded here using repr() to handle escaping ---
file_contents['funclip/__init__.py'] = ''
file_contents['funclip/introduction.py'] = repr('... full content of introduction.py ...')
file_contents['funclip/launch.py'] = repr('... full content of launch.py with debug wrappers ...')
file_contents['funclip/videoclipper.py'] = repr('... full content of videoclipper.py ...')
file_contents['funclip/llm/gemini_api.py'] = repr('... full content of gemini_api.py ...')
# ... and so on for all other .py and .json files

bootstrap_code = []
bootstrap_code.append('import os')
bootstrap_code.append('file_contents = {}')
for path, content in file_contents.items():
    bootstrap_code.append(f"file_contents['{path}'] = {content}")
bootstrap_code.append('''
for path, content in file_contents.items():
    os.makedirs(os.path.dirname(path), exist_ok=True)
    with open(path, 'w', encoding='utf-8') as f:
        f.write(content)
print('✅ Application files written.')
''')

with open("bootstrap.py", "w") as f:
    f.write('\n'.join(bootstrap_code))

!python bootstrap.py

# Step 3: Download font asset
!wget -q https://isv-data.oss-cn-hangzhou.aliyuncs.com/ics/MaaS/ClipVideo/STHeitiMedium.ttc -O font/STHeitiMedium.ttc
print("✅ Assets downloaded.")

# Step 4: Launch FunClip
print("🚀 Launching FunClip... please wait for the public URL to appear.")
!python -m funclip.launch --share
```
